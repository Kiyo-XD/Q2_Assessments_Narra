```python
def add_numbers(num1, num2):
  sum = num1 + num2
  return sum

def subtract_numbers(num1, num2):
  difference = num1 - num2
  return difference

def multiply_numbers(num1, num2):
  product = num1 * num2
  return product

def divide_numbers(num1, num2):
  if num2 == 0:
    print("Divisor cannot be 0, please try again.")
  else:
    quotient = num1 / num2
    return quotient

num1 = float(input("Enter your first number: "))
num2 = float(input("Enter your second number: "))

while True:
  operation = input("What operation(add, subtract, multiply, or divide) should be performed on these numbers? ")

  if operation == "add":
    print("The sum is", add_numbers(num1, num2))
    break

  elif operation == "subtract":
    print("The difference is", subtract_numbers(num1, num2))
    break

  elif operation == "multiply":
    print("The product is", multiply_numbers(num1, num2))
    break

  elif operation == "divide":
    print("The quotient is", divide_numbers(num1, num2))
    break

  else:
    print("This is cap sensitive, please try again.")

#What functions did you create in your calculator program? add_numbers, subtract_numbers, multiply_numbers, divide_numbers
#What parameters did your functions use? num1 and num2
#What arguments were passed when the functions were called? num1 and num2
#How did your program use the returned value? It used the returned value in the chosen operation.
#Why is it better to divide the calculator program into functions instead of writing everything in one long block of code? Because it increases readability and reusability.
```
